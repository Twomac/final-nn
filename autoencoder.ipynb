# Autoencoder for Digits Dataset

In this notebook, we train a 64x16x64 autoencoder on the sklearn digits dataset as per Step 2 of the project requirements.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from nn.nn import NeuralNetwork
```

## 1. Load and Preprocess Data

```python
digits = load_digits()
X = digits.data

# Normalize data to [0, 1] range. The digits data is 0-16 integers.
# Normalization is crucial for using sigmoid activation in the output layer.
X_normalized = X / 16.0

# Split into training and validation sets
X_train, X_val, _, _ = train_test_split(X_normalized, X_normalized, test_size=0.2, random_state=42)
```

## 2. Define Architecture and Train

```python
arch = [
    {'input_dim': 64, 'output_dim': 16, 'activation': 'relu'},
    {'input_dim': 16, 'output_dim': 64, 'activation': 'sigmoid'}
]

# Initialize the neural network
nn = NeuralNetwork(nn_arch=arch, lr=0.1, seed=42, batch_size=32, epochs=100, loss_function='mean_squared_error')

# Train the autoencoder (target is the input)
train_loss, val_loss = nn.fit(X_train, X_train, X_val, X_val)
```

## 3. Plot Loss and Evaluate

```python
plt.figure(figsize=(10, 6))
plt.plot(train_loss, label='Training Loss')
plt.plot(val_loss, label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.title('Autoencoder Training Loss')
plt.show()
```

```python
reconstructions = nn.predict(X_val)
mse = np.mean((X_val - reconstructions)**2)
print(f"Average Reconstruction Error (MSE) on Validation Set: {mse:.4f}")
```

## 4. Hyperparameter Explanation

**Architecture**: 64x16x64 as requested. ReLU is used for the encoding layer for sparsity and efficiency. Sigmoid is used for the output layer because the input data was normalized to the [0, 1] range.

**Learning Rate (0.1)**: A moderate learning rate was chosen to ensure convergence without significant oscillation.

**Batch Size (32)**: Standard batch size to balance gradient estimation accuracy with computational speed.

**Epochs (100)**: Chosen to allow sufficient passes for the loss to plateau.
